## 0. Setup

```python
import random
import warnings
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from PIL import Image
from sklearn.model_selection import train_test_split
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms
from torchvision.models import EfficientNet_B0_Weights, efficientnet_b0

warnings.filterwarnings('ignore')
random.seed(42)
np.random.seed(42)
torch.manual_seed(42)

DEVICE = torch.device(
    'cuda' if torch.cuda.is_available()
    else 'mps' if torch.backends.mps.is_available()
    else 'cpu'
)
print(f'Device: {DEVICE}')
```

```python
DATA_DIR   = Path('PokemonData')
MODEL_PATH    = Path('best_model_v2.pth')   # frozen v2 baseline — do NOT overwrite
MODEL_PATH_V3 = Path('best_model_v3.pth')   # v3 checkpoint (Tier S improvements)

IMG_SIZE         = 224
BATCH_SIZE       = 32
NUM_EPOCHS_HEAD  = 5    # Phase 1 — classifier head only
NUM_EPOCHS_FULL  = 20   # Phase 2 — max epochs (early stopping may trigger earlier)
LR_HEAD          = 1e-3
LR_FULL          = 1e-4
PATIENCE         = 7    # Phase 2 early stopping patience

IMG_EXTS = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.webp'}

def list_images(folder):
    """Return only valid image files, ignoring SVG and other non-PIL formats."""
    return [p for p in Path(folder).iterdir() if p.suffix.lower() in IMG_EXTS]
```

## 2. Data Preparation

### 🆕 v3 Change 1: Modern augmentation (RandAugment + RandomResizedCrop + RandomErasing)

**What changed:** Replaced fixed `Resize` + `RandomRotation(15)` + `ColorJitter` with the modern ImageNet recipe: `RandomResizedCrop(scale=(0.7,1.0))`, `RandAugment(2, 9)`, and `RandomErasing(p=0.25)` after `ToTensor`.

**Why:** With only ~45 images/class, weak augmentation is the dominant overfitting source — v2 train acc hit 99.7% while val plateaued at 95%. RandAugment + RandomResizedCrop are the de-facto modern baseline; RandomErasing is a cheap regularizer. *(CNN_BEST_PRACTICES.md, Tier S #1.)*

**Expected effect:** +1–3% top-1, better generalization, train/val gap shrinks.


```python
# Transforms — v3: stronger augmentation (RandAugment + RandomResizedCrop + RandomErasing)
train_transforms = transforms.Compose([
    transforms.RandomResizedCrop(IMG_SIZE, scale=(0.7, 1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.RandAugment(num_ops=2, magnitude=9),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
    transforms.RandomErasing(p=0.25),
])

eval_transforms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])
```

```python
from collections import Counter

# Load dataset three times with different transforms to avoid data leakage
train_data = datasets.ImageFolder(DATA_DIR, transform=train_transforms)
val_data   = datasets.ImageFolder(DATA_DIR, transform=eval_transforms)
test_data  = datasets.ImageFolder(DATA_DIR, transform=eval_transforms)

n_total = len(train_data)
labels  = train_data.targets  # class index for every image

# Stratified split: 70% train / 15% val / 15% test
# Guarantees each of the 150 classes is proportionally represented in all three sets
train_idx, valtest_idx = train_test_split(
    range(n_total), test_size=0.30, stratify=labels, random_state=42
)
valtest_labels = [labels[i] for i in valtest_idx]
val_idx, test_idx = train_test_split(
    valtest_idx, test_size=0.50, stratify=valtest_labels, random_state=42
)

train_set = Subset(train_data, train_idx)
val_set   = Subset(val_data,   val_idx)
test_set  = Subset(test_data,  test_idx)

print(f'Train : {len(train_set):,} images')
print(f'Val   : {len(val_set):,} images')
print(f'Test  : {len(test_set):,} images')

# Verify stratification: every class should appear in val and test
val_counts  = Counter(labels[i] for i in val_idx)
test_counts = Counter(labels[i] for i in test_idx)
print(f'Val  — min / max samples per class: {min(val_counts.values())} / {max(val_counts.values())}')
print(f'Test — min / max samples per class: {min(test_counts.values())} / {max(test_counts.values())}')
```

### 🆕 v3 Change 2: MixUp / CutMix collate

**What changed:** Added a custom `collate_mix` for `train_loader` that randomly applies MixUp (α=0.2) or CutMix (α=1.0) per batch, producing soft labels.

**Why:** Strongest regularizer available for small per-class counts (~45 imgs/class). Forces the model to learn smoother decision boundaries and directly fixes the "100% confidence on wrong predictions" pathology. *(CNN_BEST_PRACTICES.md, Tier S #1.)*

**Expected effect:** +0.5–2% top-1, dramatically better calibration. Note: train-batch labels are now soft, so `train_one_epoch` uses `labels.argmax(1)` for accuracy.


```python
# v3: MixUp / CutMix collate — applies to train_loader only
from torchvision.transforms import v2 as tvv2

cutmix_or_mixup = tvv2.RandomChoice([
    tvv2.MixUp(alpha=0.2, num_classes=NUM_CLASSES),
    tvv2.CutMix(alpha=1.0, num_classes=NUM_CLASSES),
])

def collate_mix(batch):
    imgs, labels = torch.utils.data.default_collate(batch)
    return cutmix_or_mixup(imgs, labels)
```

```python
train_loader = DataLoader(train_set, batch_size=BATCH_SIZE, shuffle=True,  num_workers=2, pin_memory=True, collate_fn=collate_mix)
val_loader   = DataLoader(val_set,   batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)
test_loader  = DataLoader(test_set,  batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

# Quick sanity check
imgs, labels = next(iter(train_loader))
print(f'Batch shape : {imgs.shape}  — expected [32, 3, 224, 224]')
# With MixUp/CutMix, labels are soft (one-hot mixed) -> show shape instead of range
print(f'Label shape : {tuple(labels.shape)}  — soft labels from MixUp/CutMix')
```

## 3. Model Training

We use **EfficientNet-B0** pretrained on ImageNet and adapt the classifier head for 150 Pokémon classes.

Training has two phases:
- **Phase 1** — freeze the backbone and train only the new classifier head (5 epochs)
- **Phase 2** — unfreeze everything and fine-tune with a lower learning rate (up to 20 epochs, early stopping patience = 7)

```python
model = efficientnet_b0(weights=EfficientNet_B0_Weights.IMAGENET1K_V1)
model.classifier[1] = nn.Linear(model.classifier[1].in_features, NUM_CLASSES)
model = model.to(DEVICE)

print(f'Classifier head: {model.classifier[1]}')
trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f'Trainable parameters: {trainable:,}')
```

### 🆕 v3 Change 5: Test-Time Augmentation (TTA) + soft-label-aware training step

**What changed:** `train_one_epoch` now handles MixUp's soft labels (uses `argmax` of the mixed target for the train-accuracy proxy). Added new `tta_evaluate` that averages softmax over the original image and its horizontal flip.

**Why:** MixUp produces 2-D labels — the old `.sum() == labels` would crash. TTA is a free +0.5–1.5% top-1 with zero retraining cost. *(CNN_BEST_PRACTICES.md, Tier S #4.)*

**Expected effect:** Training loop stays compatible with MixUp; final test top-1 gets a small TTA boost on top of the trained gains.


```python
def train_one_epoch(model, loader, optimizer, criterion, scaler=None):
    """v3: supports soft labels (MixUp/CutMix) and optional AMP scaler."""
    model.train()
    total_loss, correct, total = 0.0, 0, 0
    for imgs, labels in loader:
        imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        # Soft labels (MixUp) -> use argmax over the one-hot mix for a rough train-acc proxy
        target_idx = labels.argmax(1) if labels.ndim == 2 else labels
        total_loss += loss.item() * imgs.size(0)
        correct    += (outputs.argmax(1) == target_idx).sum().item()
        total      += imgs.size(0)
    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    for imgs, labels in loader:
        imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
        outputs = model(imgs)
        loss = criterion(outputs, labels)
        total_loss += loss.item() * imgs.size(0)
        correct    += (outputs.argmax(1) == labels).sum().item()
        total      += imgs.size(0)
    return total_loss / total, correct / total


@torch.no_grad()
def tta_evaluate(model, loader):
    """v3: Test-Time Augmentation — averages softmax over original + horizontal flip."""
    model.eval()
    correct, total = 0, 0
    for imgs, labels in loader:
        imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
        probs = model(imgs).softmax(1) + model(torch.flip(imgs, dims=[3])).softmax(1)
        correct += (probs.argmax(1) == labels).sum().item()
        total   += imgs.size(0)
    return correct / total
```

### 🆕 v3 Changes 3 & 4 (Phase 1): Label smoothing + AdamW

**What changed:** `CrossEntropyLoss()` → `CrossEntropyLoss(label_smoothing=0.1)`. `Adam` → `AdamW(weight_decay=1e-4)`. Checkpoint now saved to `best_model_v3.pth` so the v2 baseline file is preserved.

**Why:**
- Label smoothing directly attacks v2's miscalibration ("100% confidence on wrong predictions"). *(Tier S #2.)*
- `Adam`'s L2 regularization is mathematically broken when combined with adaptive learning rates; `AdamW` decouples weight decay and is the modern default. *(Tier S #3.)*

**Expected effect:** Better-calibrated probabilities, modest val-acc bump.


```python
# v3: label-smoothed CE + AdamW with weight decay
criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
history   = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': []}
best_val_acc = 0.0

# ── Phase 1: Train classifier head only ──────────────────────────────────────
for param in model.features.parameters():
    param.requires_grad = False

optimizer = optim.AdamW(model.classifier.parameters(), lr=LR_HEAD, weight_decay=1e-4)

print('Phase 1 — Classifier head only')
for epoch in range(NUM_EPOCHS_HEAD):
    tr_loss, tr_acc = train_one_epoch(model, train_loader, optimizer, criterion)
    vl_loss, vl_acc = evaluate(model, val_loader, criterion)
    history['train_loss'].append(tr_loss)
    history['val_loss'].append(vl_loss)
    history['train_acc'].append(tr_acc)
    history['val_acc'].append(vl_acc)
    if vl_acc > best_val_acc:
        best_val_acc = vl_acc
        torch.save(model.state_dict(), MODEL_PATH_V3)
    print(f'  Epoch {epoch+1}/{NUM_EPOCHS_HEAD} | '
          f'train loss {tr_loss:.4f} acc {tr_acc:.3f} | '
          f'val loss {vl_loss:.4f} acc {vl_acc:.3f}')
```

### 🆕 v3 Change 4 (Phase 2): AdamW + linear warmup + cosine

**What changed:** Phase-2 optimizer is `AdamW(weight_decay=1e-4)`. Scheduler is now `SequentialLR(LinearLR start_factor=0.1 for 2 epochs, then CosineAnnealingLR)`. LR is logged each epoch.

**Why:** When the backbone is unfrozen with a fresh head, the initial gradients can wreck pretrained features. Linear warmup ramps the LR from 1e-5 → 1e-4 over 2 epochs, then cosine anneals. *(CNN_BEST_PRACTICES.md, Tier S #3 — "critical when unfreezing the backbone".)*

**Expected effect:** Steadier early training, slightly higher final val accuracy.


```python
# ── Phase 2: Fine-tune all layers ─────────────────────────────────────────────
for param in model.features.parameters():
    param.requires_grad = True

# v3: AdamW + linear warmup (2 epochs) then cosine annealing over the remaining epochs
optimizer = optim.AdamW(model.parameters(), lr=LR_FULL, weight_decay=1e-4)
warmup    = optim.lr_scheduler.LinearLR(optimizer, start_factor=0.1, total_iters=2)
cosine    = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=max(1, NUM_EPOCHS_FULL - 2))
scheduler = optim.lr_scheduler.SequentialLR(optimizer, [warmup, cosine], milestones=[2])
patience_counter = 0

print(f'Phase 2 — Full fine-tuning  (early stopping patience={PATIENCE})')
for epoch in range(NUM_EPOCHS_FULL):
    tr_loss, tr_acc = train_one_epoch(model, train_loader, optimizer, criterion)
    vl_loss, vl_acc = evaluate(model, val_loader, criterion)
    scheduler.step()
    history['train_loss'].append(tr_loss)
    history['val_loss'].append(vl_loss)
    history['train_acc'].append(tr_acc)
    history['val_acc'].append(vl_acc)

    if vl_acc > best_val_acc:
        best_val_acc     = vl_acc
        patience_counter = 0
        torch.save(model.state_dict(), MODEL_PATH_V3)
        tag = '  ← best'
    else:
        patience_counter += 1
        tag = f'  (no improve {patience_counter}/{PATIENCE})'

    print(f'  Epoch {epoch+1:2d}/{NUM_EPOCHS_FULL} | '
          f'train loss {tr_loss:.4f} acc {tr_acc:.3f} | '
          f'val loss {vl_loss:.4f} acc {vl_acc:.3f}  '
          f'lr={optimizer.param_groups[0]["lr"]:.2e}{tag}')

    if patience_counter >= PATIENCE:
        print(f'\nEarly stopping triggered — no improvement for {PATIENCE} epochs.')
        break

print(f'\nBest val accuracy: {best_val_acc:.4f}')
```

```python
# Training curves
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
phase_boundary = NUM_EPOCHS_HEAD - 0.5

for ax, metric, title in zip(axes, [('train_loss','val_loss'), ('train_acc','val_acc')],
                              ['Loss', 'Accuracy']):
    ax.plot(history[metric[0]], label='Train')
    ax.plot(history[metric[1]], label='Val')
    ax.axvline(phase_boundary, color='gray', linestyle='--', label='Phase 2 start')
    ax.set_title(title)
    ax.set_xlabel('Epoch')
    ax.legend()

plt.tight_layout()
plt.show()
```

## 4. Evaluation

```python
# Load best checkpoint
model.load_state_dict(torch.load(MODEL_PATH_V3, map_location=DEVICE))

_, test_acc = evaluate(model, test_loader, criterion)
print(f'Top-1 accuracy: {test_acc:.4f} ({test_acc*100:.2f}%)')

@torch.no_grad()
def top5_accuracy(model, loader):
    model.eval()
    correct, total = 0, 0
    for imgs, labels in loader:
        imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
        top5 = model(imgs).topk(5, dim=1).indices
        correct += (top5 == labels.unsqueeze(1)).any(dim=1).sum().item()
        total   += imgs.size(0)
    return correct / total

top5 = top5_accuracy(model, test_loader)
print(f'Top-5 accuracy: {top5:.4f} ({top5*100:.2f}%)')
```

```python
# Per-class accuracy
@torch.no_grad()
def per_class_accuracy(model, loader):
    model.eval()
    correct = torch.zeros(NUM_CLASSES)
    total   = torch.zeros(NUM_CLASSES)
    for imgs, labels in loader:
        imgs, labels = imgs.to(DEVICE), labels.to(DEVICE)
        preds = model(imgs).argmax(1)
        for c in range(NUM_CLASSES):
            mask = labels == c
            correct[c] += (preds[mask] == c).sum().item()
            total[c]   += mask.sum().item()
    return (correct / total.clamp(min=1)).numpy()

class_acc  = per_class_accuracy(model, test_loader)
sorted_idx = np.argsort(class_acc)

fig, ax = plt.subplots(figsize=(22, 4))
ax.bar(range(NUM_CLASSES), class_acc[sorted_idx], color='steelblue')
ax.set_xticks(range(NUM_CLASSES))
ax.set_xticklabels([class_names[i] for i in sorted_idx], rotation=90, fontsize=6)
ax.axhline(test_acc, color='crimson', linestyle='--', label=f'Overall = {test_acc:.2f}')
ax.set_title('Per-class accuracy (sorted)')
ax.set_ylabel('Accuracy')
ax.legend()
plt.tight_layout()
plt.show()

print('5 hardest classes:')
for i in sorted_idx[:5]:
    print(f'  {class_names[i]:25s} {class_acc[i]:.2f}')
```

```python
# Sample correct and incorrect predictions
@torch.no_grad()
def collect_predictions(model, loader, max_batches=6):
    model.eval()
    all_imgs, all_preds, all_labels = [], [], []
    for i, (imgs, labels) in enumerate(loader):
        preds = model(imgs.to(DEVICE)).argmax(1).cpu()
        all_imgs.append(imgs)
        all_preds.append(preds)
        all_labels.append(labels)
        if i + 1 >= max_batches:
            break
    return torch.cat(all_imgs), torch.cat(all_preds), torch.cat(all_labels)

imgs_t, preds_t, labels_t = collect_predictions(model, test_loader)

mean = torch.tensor([0.485, 0.456, 0.406]).view(3,1,1)
std  = torch.tensor([0.229, 0.224, 0.225]).view(3,1,1)
denorm = lambda t: (t * std + mean).clamp(0, 1).permute(1, 2, 0).numpy()

correct_idx   = (preds_t == labels_t).nonzero(as_tuple=True)[0][:8]
incorrect_idx = (preds_t != labels_t).nonzero(as_tuple=True)[0][:8]

fig, axes = plt.subplots(2, 8, figsize=(20, 6))
for ax, i in zip(axes[0], correct_idx):
    ax.imshow(denorm(imgs_t[i]))
    ax.set_title(class_names[preds_t[i]], fontsize=7, color='green')
    ax.axis('off')
for ax, i in zip(axes[1], incorrect_idx):
    ax.imshow(denorm(imgs_t[i]))
    ax.set_title(f'Pred: {class_names[preds_t[i]]}\nTrue: {class_names[labels_t[i]]}',
                 fontsize=6, color='red')
    ax.axis('off')

axes[0][0].set_ylabel('Correct', fontsize=10)
axes[1][0].set_ylabel('Incorrect', fontsize=10)
plt.suptitle('Sample predictions', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix

# Load best checkpoint (safe to run without retraining)
model.load_state_dict(torch.load(MODEL_PATH_V3, map_location=DEVICE))
model.eval()

# Collect full test set predictions
@torch.no_grad()
def get_all_predictions(model, loader):
    all_preds, all_labels = [], []
    for imgs, labels in loader:
        preds = model(imgs.to(DEVICE)).argmax(1).cpu()
        all_preds.append(preds)
        all_labels.append(labels)
    return torch.cat(all_preds).numpy(), torch.cat(all_labels).numpy()

all_preds, all_labels = get_all_predictions(model, test_loader)
cm_full = confusion_matrix(all_labels, all_preds)

# Top 10 confused pairs (off-diagonal only)
cm_nodiag = cm_full.copy()
np.fill_diagonal(cm_nodiag, 0)
confused_pairs = [
    (cm_nodiag[i, j], class_names[i], class_names[j])
    for i in range(NUM_CLASSES)
    for j in range(NUM_CLASSES)
    if cm_nodiag[i, j] > 0
]
print('Top 10 most confused pairs (true → predicted):')
for count, true_cls, pred_cls in sorted(confused_pairs, reverse=True)[:10]:
    print(f'  {count}x  {true_cls:22s} → {pred_cls}')

# Compute per-class accuracy to find the 20 hardest classes
class_correct = np.zeros(NUM_CLASSES)
class_total   = np.zeros(NUM_CLASSES)
for pred, label in zip(all_preds, all_labels):
    class_total[label]   += 1
    class_correct[label] += (pred == label)
class_acc_cm = class_correct / np.maximum(class_total, 1)
hard_idx     = np.argsort(class_acc_cm)[:20]

# Heatmap focused on the 20 hardest classes
focus_cm    = cm_full[np.ix_(hard_idx, hard_idx)]
focus_names = [class_names[i] for i in hard_idx]

fig, ax = plt.subplots(figsize=(12, 10))
im = ax.imshow(focus_cm, cmap='Blues')
ax.set_xticks(range(20))
ax.set_yticks(range(20))
ax.set_xticklabels(focus_names, rotation=90, fontsize=8)
ax.set_yticklabels(focus_names, fontsize=8)
ax.set_xlabel('Predicted')
ax.set_ylabel('True')
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.show()
```

```python
def recommend_sixth(team_of_5, top_n=5):
    """
    Recommend the best 6th Pokemon to complete a team of 5.
    Scoring: number of new unique types the candidate adds to the team.
    """
    assert len(team_of_5) == 5, 'Team must have exactly 5 Pokemon.'

    team_types = set()
    for pokemon in team_of_5:
        team_types.update(POKEMON_TYPES.get(pokemon, []))

    scores = {}
    for pokemon, types in POKEMON_TYPES.items():
        if pokemon in team_of_5:
            continue
        scores[pokemon] = len(set(types) - team_types)

    ranked = sorted(scores.items(), key=lambda x: (-x[1], x[0]))

    print(f'Team         : {team_of_5}')
    print(f'Types covered: {sorted(team_types)}')
    print(f'\nTop {top_n} recommendations for the 6th slot:')
    for rank, (pokemon, score) in enumerate(ranked[:top_n], 1):
        new_types = sorted(set(POKEMON_TYPES[pokemon]) - team_types)
        print(f'  {rank}. {pokemon:22s} adds {score} new type(s): {new_types}')

    return [p for p, _ in ranked[:top_n]]
```

```python
def classify_image(img_path, model, transform):
    """Run inference on one image and return the predicted Pokemon name."""
    model.eval()
    img    = Image.open(img_path).convert('RGB')
    tensor = transform(img).unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        probs   = torch.softmax(model(tensor), dim=1)[0]
        top5p, top5i = probs.topk(5)
    print(f'Image: {Path(img_path).name}')
    for p, i in zip(top5p, top5i):
        print(f'  {idx_to_class[i.item()]:22s} {p.item()*100:.1f}%')
    return idx_to_class[top5i[0].item()]
```

## 6. v3 vs v2 — Head-to-Head Comparison

To know whether the Tier S improvements *actually* helped (vs noise from a small test set with ~7 images/class), we compare v2 and v3 on the same test set using:

1. **Metrics table** — top-1, top-1 + TTA, top-5, loss, **Expected Calibration Error (ECE)**, mean confidence on wrong predictions.
2. **McNemar's paired test** — is the accuracy gap statistically real?
3. **Per-class regression check** — did MixUp hurt any specific class?
4. **Reliability diagram** — visual calibration check.

**Decision rule:** v3 is declared an improvement iff McNemar p < 0.05 in favor of v3, **and** ECE(v3) < ECE(v2), **and** no more than 5 classes regress by > 10 pp.


```python
# Cell A — Load both models on the SAME test set
def build_model():
    m = efficientnet_b0(weights=None)
    m.classifier[1] = nn.Linear(m.classifier[1].in_features, NUM_CLASSES)
    return m.to(DEVICE)

model_v2 = build_model()
model_v2.load_state_dict(torch.load(MODEL_PATH, map_location=DEVICE))
model_v2.eval()

model_v3 = build_model()
model_v3.load_state_dict(torch.load(MODEL_PATH_V3, map_location=DEVICE))
model_v3.eval()

print(f'v2 weights: {MODEL_PATH}')
print(f'v3 weights: {MODEL_PATH_V3}')
```

```python
# Cell B — Collect predictions + per-image softmax for both models
@torch.no_grad()
def collect(model, loader, tta=False):
    model.eval()
    all_probs, all_labels = [], []
    for imgs, labels in loader:
        imgs = imgs.to(DEVICE)
        probs = model(imgs).softmax(1)
        if tta:
            probs = (probs + model(torch.flip(imgs, dims=[3])).softmax(1)) / 2
        all_probs.append(probs.cpu())
        all_labels.append(labels)
    return torch.cat(all_probs), torch.cat(all_labels)

probs_v2,  labels_test = collect(model_v2, test_loader)
probs_v3,  _           = collect(model_v3, test_loader)
probs_v3_tta, _        = collect(model_v3, test_loader, tta=True)

preds_v2     = probs_v2.argmax(1)
preds_v3     = probs_v3.argmax(1)
preds_v3_tta = probs_v3_tta.argmax(1)

correct_v2     = (preds_v2     == labels_test).numpy()
correct_v3     = (preds_v3     == labels_test).numpy()
correct_v3_tta = (preds_v3_tta == labels_test).numpy()

print(f'v2     top-1: {correct_v2.mean():.4f}')
print(f'v3     top-1: {correct_v3.mean():.4f}')
print(f'v3+TTA top-1: {correct_v3_tta.mean():.4f}')
```

```python
# Cell C — Metrics table (top-1, top-5, loss, ECE, mean wrong-confidence)
import pandas as pd

def top5(probs, labels):
    top5_idx = probs.topk(5, dim=1).indices
    return (top5_idx == labels.unsqueeze(1)).any(dim=1).float().mean().item()

def ece(probs, labels, n_bins=15):
    confs, preds = probs.max(1)
    accs = (preds == labels).float()
    bins = torch.linspace(0, 1, n_bins + 1)
    e = 0.0
    for lo, hi in zip(bins[:-1], bins[1:]):
        mask = (confs > lo) & (confs <= hi)
        if mask.sum() > 0:
            e += (mask.float().mean() * (accs[mask].mean() - confs[mask].mean()).abs()).item()
    return e

def mean_wrong_conf(probs, labels):
    confs, preds = probs.max(1)
    wrong = preds != labels
    return confs[wrong].mean().item() if wrong.any() else float('nan')

def ce_loss(probs, labels):
    return -torch.log(probs[torch.arange(len(labels)), labels].clamp(min=1e-12)).mean().item()

rows = []
for name, probs in [('v2', probs_v2), ('v3', probs_v3), ('v3 + TTA', probs_v3_tta)]:
    preds = probs.argmax(1)
    rows.append({
        'model':            name,
        'top-1':            (preds == labels_test).float().mean().item(),
        'top-5':            top5(probs, labels_test),
        'CE loss':          ce_loss(probs, labels_test),
        'ECE (15-bin)':     ece(probs, labels_test),
        'mean conf (wrong)': mean_wrong_conf(probs, labels_test),
    })

df = pd.DataFrame(rows).set_index('model')
df.loc['Δ (v3 - v2)']     = df.loc['v3']       - df.loc['v2']
df.loc['Δ (v3+TTA - v2)'] = df.loc['v3 + TTA'] - df.loc['v2']
df.round(4)
```

```python
# Cell D — McNemar's paired test (v2 vs v3, same test images)
from statsmodels.stats.contingency_tables import mcnemar

both_correct = int(( correct_v2 &  correct_v3).sum())
v2_only      = int(( correct_v2 & ~correct_v3).sum())
v3_only      = int((~correct_v2 &  correct_v3).sum())
both_wrong   = int((~correct_v2 & ~correct_v3).sum())

table = [[both_correct, v2_only],
         [v3_only,      both_wrong]]

print('Contingency table:')
print(f'                v3 correct   v3 wrong')
print(f'v2 correct      {both_correct:>6d}      {v2_only:>6d}')
print(f'v2 wrong        {v3_only:>6d}      {both_wrong:>6d}')

# Exact binomial McNemar when discordant counts are small
result = mcnemar(table, exact=(v2_only + v3_only) < 25)
print(f'\nMcNemar statistic: {result.statistic:.3f}')
print(f'p-value          : {result.pvalue:.4g}')

direction = 'v3 better' if v3_only > v2_only else 'v2 better' if v2_only > v3_only else 'tie'
print(f'Direction        : {direction}  (v3-only correct: {v3_only}, v2-only correct: {v2_only})')
```

```python
# Cell E — Per-class accuracy regression check
acc_v2_per = np.zeros(NUM_CLASSES)
acc_v3_per = np.zeros(NUM_CLASSES)
cnt_per    = np.zeros(NUM_CLASSES)
for i, l in enumerate(labels_test.numpy()):
    cnt_per[l] += 1
    acc_v2_per[l] += correct_v2[i]
    acc_v3_per[l] += correct_v3[i]
acc_v2_per /= np.maximum(cnt_per, 1)
acc_v3_per /= np.maximum(cnt_per, 1)
diff = acc_v3_per - acc_v2_per
order = np.argsort(diff)

fig, ax = plt.subplots(figsize=(22, 4))
colors = ['crimson' if d < 0 else 'steelblue' for d in diff[order]]
ax.bar(range(NUM_CLASSES), diff[order], color=colors)
ax.axhline(0, color='black', linewidth=0.5)
ax.set_xticks(range(NUM_CLASSES))
ax.set_xticklabels([class_names[i] for i in order], rotation=90, fontsize=6)
ax.set_ylabel('v3 - v2 per-class accuracy')
ax.set_title('Per-class accuracy change (red = v3 regressed)')
plt.tight_layout()
plt.show()

regressed = [(class_names[i], diff[i]) for i in range(NUM_CLASSES) if diff[i] < -0.10]
print(f'\nClasses regressed by > 10pp: {len(regressed)}')
for name, d in sorted(regressed, key=lambda x: x[1])[:10]:
    print(f'  {name:25s} {d:+.2f}')
```

```python
# Cell F — Reliability diagram (calibration)
def reliability(probs, labels, n_bins=10):
    confs, preds = probs.max(1)
    accs = (preds == labels).float()
    bins = torch.linspace(0, 1, n_bins + 1)
    xs, ys, ws = [], [], []
    for lo, hi in zip(bins[:-1], bins[1:]):
        mask = (confs > lo) & (confs <= hi)
        if mask.sum() > 0:
            xs.append(confs[mask].mean().item())
            ys.append(accs[mask].mean().item())
            ws.append(mask.sum().item())
    return xs, ys, ws

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, (name, probs) in zip(axes, [('v2', probs_v2), ('v3 + TTA', probs_v3_tta)]):
    xs, ys, ws = reliability(probs, labels_test)
    ax.plot([0, 1], [0, 1], 'k--', alpha=0.5, label='perfect calibration')
    ax.scatter(xs, ys, s=[max(20, w*3) for w in ws], alpha=0.7)
    ax.plot(xs, ys, 'o-')
    ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.set_xlabel('Confidence (max softmax)')
    ax.set_ylabel('Empirical accuracy')
    ax.set_title(f'Reliability diagram — {name}')
    ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Cell G — Verdict
ece_v2 = ece(probs_v2,     labels_test)
ece_v3 = ece(probs_v3_tta, labels_test)
n_regressed = sum(1 for d in diff if d < -0.10)
mcnemar_pvalue = result.pvalue
v3_won = v3_only > v2_only

cond1 = (mcnemar_pvalue < 0.05) and v3_won
cond2 = ece_v3 < ece_v2
cond3 = n_regressed <= 5

print('=' * 60)
print('VERDICT  (v3 vs v2)')
print('=' * 60)
print(f'  [{"✓" if cond1 else "✗"}] McNemar p < 0.05 in favor of v3   (p = {mcnemar_pvalue:.4g})')
print(f'  [{"✓" if cond2 else "✗"}] ECE(v3+TTA) < ECE(v2)             ({ece_v3:.4f} vs {ece_v2:.4f})')
print(f'  [{"✓" if cond3 else "✗"}] ≤ 5 classes regressed > 10pp      ({n_regressed} regressed)')
print('-' * 60)
print(f'  Overall: {"✅ v3 is an improvement" if (cond1 and cond2 and cond3) else "⚠️ inconclusive — review failing condition"}')
```
